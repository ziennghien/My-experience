**Có sử dụng pretrain model: phoBERT**

```python
# Use a pipeline as a high-level helper
import torch
from transformers import pipeline

pipe = pipeline("text-classification", model="UchihaMadara/phobert-finetuned-sentiment-analysis")

# Load model directly
from transformers import AutoTokenizer, AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained("UchihaMadara/phobert-finetuned-sentiment-analysis")
model = AutoModelForSequenceClassification.from_pretrained("UchihaMadara/phobert-finetuned-sentiment-analysis")

model.eval()  # Đưa mô hình về chế độ đánh giá

# Danh sách nhãn của mô hình
LABELS = ["negative", "neutral", "positive"]

def classify_text(text):
    '''
    Phân loại comment thành positive, neutral hoặc negative.
    '''
    # Tokenize input
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)

    # Dự đoán
    with torch.no_grad():
        outputs = model(**inputs)
        print(outputs.logits.softmax(dim=-1).tolist())

    # Lấy giá trị logits
    logits = outputs.logits

    # Chuyển thành nhãn dự đoán
    predicted_class = torch.argmax(logits, dim=1).item()

    return LABELS[predicted_class]
```

```python
text_1 = "tui thấy bài này cũng cũng"
text_2 = "bài này mà cũng ra mắt được"
text_3 = "ai chê chứ tui thấy cũng hay mà"

print(f"Text: {text_1} → Class: {classify_text(text_1)}")
print(f"Text: {text_2} → Class: {classify_text(text_2)}")
print(f"Text: {text_3} → Class: {classify_text(text_3)}")
```
